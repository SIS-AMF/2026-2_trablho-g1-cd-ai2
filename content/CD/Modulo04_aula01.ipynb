 "Mapa Logístico" é uma equação simples usada para modelar crescimento populacional, mas que demonstra comportamento caótico. A fórmula é:

 $$x_{n+1} = r \cdot x_n \cdot (1 - x_n)$$

 Onde: $x_n$ é a população atual ($0$ é extinção, $1$ é a capacidade máxima do ecossistema). $r$ é a taxa de crescimento. Se $r$ for alto o suficiente (ex: 3.9), o sistema entra em caos.

```python
import matplotlib.pyplot as plt
import numpy as np

def mapa_logistico(r, x):
    return r * x * (1 - x)

# Parâmetros
r = 3.9
passos = 100

# Dois estados iniciais EXTREMAMENTE próximos
x1 = 0.5
x2 = 0.50000001  # Diferença de apenas 0.00000001

# Listas para guardar a história
historia_x1 = [x1]
historia_x2 = [x2]

# Simulação temporal
for _ in range(passos):
    x1 = mapa_logistico(r, x1)
    x2 = mapa_logistico(r, x2)
    historia_x1.append(x1)
    historia_x2.append(x2)

# Visualização
plt.figure(figsize=(12, 6))
plt.plot(historia_x1, label='Simulação A (Início: 0.5)', alpha=0.7)
plt.plot(historia_x2, label='Simulação B (Início: 0.50000001)', alpha=0.7, linestyle='--')
plt.title(f'Dependência Sensível das Condições Iniciais (r={r})')
plt.ylabel('Valor de X')
plt.xlabel('Tempo (Iterações)')
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)

```

### Definição Matemática de Probabilidade  
**(Abordagem Axiomática de Kolmogorov – 1933)**

A probabilidade é definida rigorosamente por meio de três axiomas sobre um **espaço de probabilidade** $(\Omega, \mathcal{F}, P)$.

#### 1. Espaço Amostral $(\Omega)$
- Conjunto de **todos os resultados possíveis** de um experimento aleatório.  
- Pode ser finito
#### 2. σ-álgebra (ou álgebra de eventos, o que podemos monitorar) $\mathcal{F}$
- Coleção de subconjuntos de $\Omega$ que chamamos de **eventos** (A).  
- Deve satisfazer:
  1. $\Omega \in \mathcal{F}$
  2. Se $A \in \mathcal{F} \Rightarrow A^c \in \mathcal{F}$
  3. Se $A_1, A_2, \dots \in \mathcal{F} \Rightarrow \bigcup_{i=1}^\infty A_i \in \mathcal{F}$

Pensa numa lista de eventos permitidos. Essa lista tem que seguir 3 regras simples:

| Regra     | Nome técnico                          | Por quê? (intuição)                                      |
|-----------------------------------------------|---------------------------------------|----------------------------------------------------------|
| 1. “Vai acontecer alguma coisa” tem que estar na lista | $\Omega \in \mathcal{F}$              | Tem que poder falar do experimento inteiro              |
| 2. Se você pode falar de um evento, pode falar do contrário | Se $A \in \mathcal{F} \Rightarrow A^c \in \mathcal{F}$ | Se posso falar “par”, tenho que poder falar “ímpar”      |
| 3. Você pode juntar infinitos eventos da lista e ainda tem que estar na lista | União contável fechada                | Se posso falar de cada dia do ano, posso falar “algum dia” |

#### Analogia

| Conceito               | Analogia do dia a dia                                      |
|------------------------|------------------------------------------------------------|
| $\Omega$               | Todas as roupas do seu armário                             |
| Evento                 | “Todas as camisetas vermelhas”                             |
| σ-álgebra $\mathcal{F}$| A lista de combinações que você realmente consegue separar (ex: por cor, por manga, etc.) |
| Probabilidade $P$      | “Qual a chance de eu pegar uma camiseta vermelha hoje?”    |

#### 3. Medida de Probabilidade $P$
Função $P: \mathcal{F} \to [0,1]$ que satisfaz os **três axiomas de Kolmogorov**:

$$
\boxed{
\begin{aligned}
&\text{1. Não-negatividade:} & P(A) \geq 0 \\
&\text{2. Normalização:}     & P(\Omega) = 1 \\
&\text{3. Aditividade:} &
\text{Se } A_1,A_2,\dots \text{ mutuamente exclusivos} \Rightarrow \\
& & P\left(\bigcup_{i=1}^\infty A_i\right) = \sum_{i=1}^\infty P(A_i)
\end{aligned}
}
$$



#### Probabilidade Condicional (definição formal)

$$
\boxed{
P(B \mid A) = \frac{P(A \cap B)}{P(A)} \quad \text{se } P(A) > 0
}
$$

#### Teorema de Bayes

$$
\boxed{
P(A \mid B) = \frac{P(B \mid A) P(A)}{P(B)}
}
$$

#### Independência (definição axiomática)

Dois eventos $A$ e $B$ são independentes se e somente se

$$
\boxed{
P(A \cap B) = P(A) \cdot P(B)
}
$$

# Distribuição Binomial

É a distribuições de probabilidade discreta Fundamental. Ela modela o número de "sucessos" `*k*` em um número fixo de tentativas ou ensaios independentes `n`, onde cada tentativa tem exatamente dois resultados possíveis (sucesso ou falha) e a probabilidade de sucesso `p` é constante em todas as tentativas.

A fórmula para calcular a probabilidade de obter exatamente `k` sucessos em `n` tentativas é dada pela Função de Probabilidade (PMF):

$$
P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}
$$

Onde:
- $ n $ = número de tentativas (ex: 10 lançamentos de moeda)
- $ k $ = número de sucessos desejados (ex: exatamente 6 caras)
- $ p $ = probabilidade de sucesso em cada tentativa
- $ \binom{n}{k} $ = coeficiente binomial = $ \frac{n!}{k!(n-k)!} $

Vamos fazer um exemplo: Uma fabrica produz uma mercadoria com defeito a cada 20 fabricadas, qual a probabilidade de encontrar em uma amostra de 100 peças pelo menos 7 peças com defeito; encontra pelo menos 10 com defeito?

Note que vamos utilizar os valores acumulados de probabilidade (CDF) para encontrar o valor de pelo menos n preças com defeito.

$$
P(X \leq k) = \sum_{i=0}^{k} \binom{n}{i} p^i (1-p)^{n-i}
$$

```python
from scipy.stats import binom
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Parâmetros do problema
n = 100  # Tamanho da amostra
p = 1/20  # Probabilidade de uma peça ter defeito (1 a cada 20)



# Calcula as probabilidades PMF (Probability Mass Function) para a distribuição binomial
# Defini o range de valores de 'k' (número de peças com defeito) para plotar
# Vamos plotar de 0 até um valor razoável
k_max_plot = 15
k_valores = np.arange(0, k_max_plot + 1)
prob_binom = binom.pmf(k=k_valores, n=n, p=p)


plt.figure(figsize=(12, 7))
sns.barplot(
    x=k_valores,
    y=prob_binom,
    color='#003366', # Cor única para a binomial
    alpha=0.9
)

plt.title('Distribuição Binomial (n=100, p=0.05)', fontsize=16)
plt.xlabel('Número de Peças com Defeito (k)', fontsize=12)
plt.ylabel('Probabilidade P(X = k)', fontsize=12)
plt.xticks(k_valores) # Garantir que todos os k_valores sejam exibidos no eixo X
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()



# Probabilidade de encontrar pelo menos 7 peças com defeito (P(X >= 7))
# É mais fácil calcular 1 - P(X <= 6)
prob_menos_que_7 = binom.cdf(k=6, n=n, p=p) # cumulative
prob_pelo_menos_7_binom = 1 - prob_menos_que_7


print(" Pelo menos 7 peças com defeito;")
print(f"Probabilidade (Binomial): {prob_pelo_menos_7_binom:.4f}\n")


#  Probabilidade de encontrar pelo menos 10 peças com defeito (P(X >= 10))
# É mais fácil calcular 1 - P(X <= 9)
prob_menos_que_10 = binom.cdf(k=9, n=n, p=p)
prob_pelo_menos_10_binom = 1 - prob_menos_que_10


print("Pelo menos 10 peças com defeito:")
print(f"Probabilidade (Binomial): {prob_pelo_menos_10_binom:.4f}")

```

# Distribuição de Poisson

Modelo clássico para contar **eventos raros e independentes** em um intervalo fixo de tempo ou espaço.

### Exemplos típicos
- Número de chamadas recebidas num call center por hora  
- Quantidade de acidentes num cruzamento por dia  
- Número de e-mails recebidos por minuto  

### Fórmula da função de probabilidade (PMF)

$$
\boxed{
P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!}
}
$$

Onde:
- $ X $ = número de eventos (k = 0, 1, 2, 3, …)
- $ \lambda $ (lambda) = número médio de eventos no intervalo (parâmetro único da distribuição)
- $ e $ ≈ 2,71828


Vamos resolver um exemplo, temos uma media de 3 acidentes por mes em um cruzamento.



```python
from scipy.stats import poisson
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#cria distribuição de poisson para media 3
poisson_dist = poisson(mu=3)

# calcula probabilidade para exatamente 5 eventos
probability = poisson_dist.pmf(5)

#Cumulative Distribution Function (CDF). calcula o acumulado até "x" valor
#probability = poisson_dist.cdf(5) - poisson_dist.cdf(0)

print(f"probabilidade para 5 acidentes : {probability:.4f}")

k_max_plot=10
k_valores = np.arange(0, k_max_plot + 1)
amostras = poisson_dist.pmf(k_valores)

plt.figure(figsize=(12, 7))
sns.barplot(
    x=k_valores,
    y=amostras,
    color='#003366', # Cor única para a binomial
    alpha=0.9
)

plt.title('Distribuição Poisson (n=100, p=0.05)', fontsize=16)
plt.xlabel('Número de acidentes (k)', fontsize=12)
plt.ylabel('Probabilidade P(X = k)', fontsize=12)
plt.xticks(k_valores) # Garantir que todos os k_valores sejam exibidos no eixo X
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
