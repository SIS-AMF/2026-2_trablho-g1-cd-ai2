### Parâmetros úteis na importação

podemos ter problemas ao importar arquivos (separador incorreto, casas decimais com vírgula, etc.). A função `read_csv` tem argumentos para lidar com isso:

* `sep`: Define o separador de colunas (ex: `sep=';'`).
* `decimal`: Define o caractere usado para separar as casas decimais (ex: `decimal=','`).
* `encoding`: Resolve problemas de codificação de caracteres (ex: `'latin-1'`, `'utf-8'`).
* `skiprows`: Pula um número de linhas do início do arquivo.
* `na_values`: Uma lista de strings que devem ser consideradas como `NA`/`NaN` (ex: `na_values=['-999', 'sem info']`).
* `dtype`: Permite especificar o tipo de cada coluna.

# 2. Manipulação de Dados com Pandas

###Encadeamento de métodos

Para tornar o codigo mais legivel, visto que a maioria das funções de manipulação do 'pandas' são *metodos* de um DataFrame, podemos chamá-las em sequência, uma após a outra por meio de **encadeamento de metodos** (Method chaining)

**Receita de bolo em Python (com method chaining):** recipiente.acrescente("farinha", "ovo", "Acucar").bata().asse(time="45", temp="200")

 O `pandas` possui inumeros métodos, podemos descar.

| Método| Descrição |
|---|---|
| `df[['col1', 'col2']]` ou `df.drop()` | Seleciona colunas |
| `df.query()` ou `df[condição]` | Filtra linhas |
| `df.sort_values()` | Ordena o DataFrame |
| `df.assign()` ou `df['nova_col'] = ...` | Cria ou modifica colunas |
| `df.groupby()` | Agrupa o DataFrame |
| `.agg()` ou outras agregações (`.mean()`, `.sum()`) | Sumariza dados agrupados |
| `pd.merge(df1, df2, on='chave', how='left')` | Junta DataFrames |
| `df['col'].value_counts()` | Conta as ocorrencias de um elemento na coluna
|`df.explode('col')`| duplica a linha para cada elemento da lista

Vamos usar o nosso DataFrame `imdb_df` para praticar. Primeiro, vamos remover linhas com valores nulos em colunas importantes para facilitar os exemplos.

Strings dispoem de varios metodos de manipulação, vamos destacar alguns:

*   `.startswith() , .endswith() e .str.contains()` : seleciona as linhas baseado no inicio, fim o u que contenha um fragmento de str, aceita Expressões Regulares.
*   `.str.strip()`: Remove espaços em branco do início E do fim da string.
*   `.str.lstrip()`: Remove espaços em branco apenas do início
*   `.str.rstrip()`: Remove espaços em branco apenas do fim
*   `.str.replace('texto_antigo', 'texto_novo')`: Substitui todas as ocorrências de um texto por outro.
*  ` .str.split('delimitador')`: Divide a string e retorna uma lista, o argumento `expand=True` ja cria direto no dataframe as colunas.
*   `.str.extract('(regex)')`: Permite usar Expressões Regulares para extrarir informações
*   `.str.get(i)`:  Pega a palavra na posição i da lista.



PAra mais info sobre expressões regulares: [regex](https://media.datacamp.com/legacy/image/upload/v1665049611/Marketing/Blog/Regular_Expressions_Cheat_Sheet.pdf)

## Criação e modificação de colunas (variaveis, atributos)

Destaco agora a importancia de conhecer o contexto do tema, uma das principais atribuições de um cientista de dados, a Engenharia de recurso, que conciste na capacidade de obter novas variaveis (com relevancia) apartir do conjunto de dados.

````python
#Metodo direto, modificando o DataFrame original
imdb_df['lucro'] = imdb_df['receita'] - imdb_df['gastos'] # determinar o lucro do filme
imdb_df[["titulo","ano","nota_imdb","lucro"]].sort_values(by='lucro',ascending=False).head()
````

````python
# Usando .assign() (cria um novo DataFrame, ideal para encadeamento)
df_com_lucro = imdb_df.assign(
    lucro_milhoes = lambda df: (df['receita'] - df['gastos']) / 1_000_000,
    gastos_milhoes = lambda df: df['gastos'] / 1_000_000,
    receita_milhoes = lambda df: df['receita'] / 1_000_000
)

df_com_lucro[['titulo', 'lucro_milhoes']].head()
````

````python
# com aplicação de logica condicional,
# necessario a biblioteca numpy
import numpy as np

#Ex: nova variavel com condição de prejuizo
df_com_prejuizo = df_com_lucro.assign(
    teve_prejuizo = lambda df: np.where(df['lucro'] < 0, 'Sim', 'Não')
)

display(df_com_prejuizo[['titulo', 'lucro_milhoes', 'teve_prejuizo']].head())
df_com_prejuizo[['titulo', 'lucro_milhoes', 'teve_prejuizo']][df_com_prejuizo['teve_prejuizo']=="Sim"] # caso quiser ver apenas os que tiverem prejuizo
df_com_prejuizo[['titulo', 'lucro_milhoes', 'teve_prejuizo',"nota_imdb"]].query('teve_prejuizo=="Sim" and nota_imdb>8').head() # bons filmes com prejuizo
````

## Agrupamento e sumarização


*    padrão em pandas é df.groupby('coluna')['outra_coluna'].mean()
*   para múltiplas agregações: df.groupby('coluna').agg(nova_coluna=('outra_coluna', 'mean'))




````python
#Receita media por Ano dos filmes
imdb_df.groupby('ano')['receita'].mean().sort_values(ascending=False)

# deversas informações por Ano
imdb_df.groupby('ano').agg(
    nota_media=('nota_imdb', 'mean'),
    desvio_padrao_nota=('nota_imdb', 'std'),
    orcamento_medio=('gastos', 'mean'),
    numero_de_filmes=('titulo', 'count')
).sort_values('numero_de_filmes', ascending=False)
````

## Merge
juntar data frames
![exemplo pelo ID](https://statisticsglobe.com/wp-content/uploads/2021/12/join-types-python-merge-programming.png)

````python
#Vamos criar uma tabela com a nota média por Ano
nota_media_por_ano = imdb_df.groupby('ano').agg(
    nota_media_anual=('nota_imdb', 'mean')
).reset_index() # .reset_index() transforma a coluna de grupo 'ano' em uma coluna normal

display(nota_media_por_ano.head())


# Agora vamos juntar (merge) essa informação de volta ao DataFrame original
imdb_com_nota_media = pd.merge(
    imdb_df,
    nota_media_por_ano,
    on='ano', # a chave da junção
    how='left'   # tipo de join (left, right, inner, outer)
)


display(imdb_com_nota_media[['ano','titulo', 'nota_imdb', 'nota_media_anual']].head())
````

## atividade
Qual diretor tem os filmes com a melhor nota, “Quentin Tarantino” ou “Steven Spielberg”, quais outras métricas você poderia usar para ajudar a definir qual é melhor e o que pode trazer incerteza avaliação. (qual você contrataria?)

````python
sumario_final = (
    imdb_df[imdb_df['direcao'].isin(["Quentin Tarantino", "Steven Spielberg"])]
    .groupby('direcao')
    .agg(
        nota_media=('nota_imdb', 'mean'),
        desvio=('nota_imdb', 'std'),
        numero_filmes=('titulo', 'size'),
        numero_criticas=('num_criticas_publico', 'sum'),
        gasto_medio=('gastos', 'mean')
    )
    .assign( # .assign() permite criar novas colunas de forma encadeada
        relacao_nota_orcamento = lambda df: (df['nota_media'] / df['gasto_medio']) * 10000000
    )
    .reset_index()
)

display(sumario_final)
````

Uma vez que temos o formato datetime, podemos usar o acesso `.dt` que dispoe de inumeros metodos para operar as datas,m mais infos em [Pandas Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.dt.month.html)

````python
# Criando uma coluna com o número do dia da semana (Segunda=0, Domingo=6),day_name() para nome
imdb_df['dia_da_semana_num'] = imdb_df['data_lancamento'].dt.day_name()
imdb_df['dia_da_semana_num']
````

Podemos usar o `.dt.strftime`, poara formatar como quisermos a datas

`%d`: Dia do mês (ex: 01, 14, 30)

`%m`: Mês como número (ex: 05, 10, 12)

`%Y`: Ano com 4 dígitos (ex: 1994)

`%y`: Ano com 2 dígitos (ex: 94)

`%B`: Nome completo do mês

`%A`: Nome completo do dia da semana

````python
#converter o formato de interesse
imdb_df['data_lancamento'].dt.strftime('%d/%m/%Y')

````
